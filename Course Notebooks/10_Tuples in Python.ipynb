## **Constructing a Tuple**

```python
tuple_1 = (1, 2, 3)
print(tuple_1)

tuple_2 = ('a', 'b', 'c')
print(tuple_2)

tuple_3 = ('ab', 2, 3.14)
print(tuple_3)
```

## **The len() function applied on Tuple**

```python
len(tuple_1)  # 📏 Returns: 3
```

## **Indexing on Tuple**

```python
tuple_3[1]  # 🎯 Access the second element → 2
```

## **Slicing on Tuple**

```python
tuple_3[1:]  # ✂️ Slice from index 1 to the end → (2, 3.14)
```

## **The .count() Method on Tuple**

```python
tuple_4 = ('a', 'a', 'a', 7, 4)
tuple_4.count('a')  # 🔢 Returns: 3 (counts occurrences of 'a')
```

## **The .index() Method on Tuple**

```python
tuple_4.index(7)  # 🔎 Returns the first index of value 7 → 3
```

## **Tuples are Concatenable**

```python
tuple_5 = tuple_1 + tuple_2 + tuple_3 + tuple_4
print(tuple_5)
```

## **Tuples are Multipliable**

```python
tuple_6 = tuple_5 * 2  # 🔁 Repeats the tuple twice
print(tuple_6)
```

## **Tuples can contain Lists and Dictionaries**

```python
tuple_7 = (1,2,"ab",[3,4,5],{"sixth" : 6, "seventh" : 7})
print(tuple_7)
```

## **Traversing in Tuples**

```python
tuple_7[3][1]  # Access 2nd element in list inside tuple → 4
```

## **The tuple() Constructor**

```python
my_list1 = [1, 2, 3, 4, 5]
tuple_8 = tuple(my_list1)  # 🔄 Converts list to tuple
print(tuple_8)
```

## **Tuples are immutable and hence it gives an error**

```python
tuple_4[0] = 3  # ❌ Attempt to modify → gives TypeError
```
