## **Creating a List**

```python
my_list = [1, 2, 3]
print(my_list)  # 🧺 A list of integers
```

```python
my_list2 = ["string", 1, 2, "Hello", 4, 5, "Yash", "#"]
print(my_list2)  # 🎒 A list with mixed data types
```

## **len() Function applied on a List**

```python
print(len(my_list))    # 📏 Length → 3
print(len(my_list2))   # 📏 Length → 8
```

## **Indexing Operation done on a List**

```python
print(my_list[1])      # 🔢 Access second element → 2
print(my_list2[0])     # 🎯 Access first element → 'string'
```

## **Slicing Operation done on a List**

```python
print(my_list[0:2])        # ✂️ Slice from index 0 to 1 → [1, 2]
print(my_list2[::-1])      # 🔁 Reverses the list
```

## **Concatenation Operation done on a list**

```python
print(my_list)
my_list3 = [4, 5, 6, 7]
print(my_list3)

my_list4 = my_list + my_list3  # ➕ Combines both lists
print(my_list4)
```

## **Multipliability Operation performed on a List**

```python
print(my_list3)
my_list5 = my_list3 * 2  # 🔁 Duplicates the list
print(my_list5)
```

## **Lists are Mutable**

```python
my_list6 = [1, 2, 3, 4, 5, 6, 7]
my_list6[2] = 8  # ♻️ Modifying a specific index
print(my_list6)  # [1, 2, 8, 4, 5, 6, 7]
```

## **The .append() Method on Lists**

```python
my_list7 = [1, 2, 3, 4, 5, 6]
my_list7.append(9)  # 📌 Adds value at the end
print(my_list7)
```

## **The .pop() Method on Lists**

```python
my_list4.pop()  # ❌ Removes the last element (default behavior)
#index number of the element to be popped out is written inside the pop bracket 
#and if nothing is written inside the bracket then the last element which was inserted will be popped out by default.
```

```python
my_popped_list = my_list4.pop(3)  # 🎯 Pops element at index 3
print(my_popped_list)
print(my_list4)  # 🧹 List after popping
```

## **.sort() Method on Lists**

```python
my_num_list = [12, 3, 7, 19, -1]
my_alpha_list = ['a', 'o', 'e', 'z']
my_jingle = [12, 2, 'z', 'a']  # ❌ Mixed types (int + str) — sorting will raise error
```

```python
# ⚠️ Cannot use .sort() in print directly
my_num_list.sort()
print(my_num_list)

my_alpha_list.sort()
print(my_alpha_list)
```

```python
my_jingle.sort() # .sort() method cannot be applied on the list which contains both alphabets and numbers. This gives an error
print(my_jingle)
```

```python
my_num2_list = [1,9,2,8]
my_sorted_num2_list = my_num2_list.sort()
print(my_sorted_num2_list) # This will give an output as none as the output as the .sort() does not make any change directly in the variable.
```

```python
# Now if you want to use the .sort() method and you want to assign a sorted list to another variable then you have to use the following way.

my_num2_list = [1,9,2,8]
my_num2_list.sort()
my_sorted_num2_list = my_num2_list
print(my_sorted_num2_list)
```

## **The .reverse() Method on Lists**

```python
my_sorted_num2_list.reverse()
print(my_sorted_num2_list)  # [9, 8, 2, 1]
```

```python
my_sorted_num2_list[::-1]
#another trick to reverse the elements in the list.
```

## **Lists allow duplicate values**

```python
my_list = [1, 2, 3, 4, 4]  # ✅ Duplicate 4 allowed
print(my_list)
```

## **Lists can also hold the Boolean Datatypes**

```python
list_4 = [True, True, True, False, False, False]
print(list_4)
```

## **List Constructor**

```python
list_test = list(("Yash is coding"))  # Creates list of characters
list_test_2 = list(("Yash", "Suhas", "Shukla"))  # Creates list of names

print(list_test)
print(list_test_2)
```

## **The type() function on List**

```python
print(type(list_4))  # Returns: <class 'list'>
```

## **The len() function on List**

```python
len(list_4)  # 📏 Length → 6
```
