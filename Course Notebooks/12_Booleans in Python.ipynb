## **The Boolean Output**

### **The "True" keyword**

```python
type(True)   # ✅ Always capitalize 'True'. Returns: <class 'bool'>
```

### **The "False" keyword**

```python
type(False)  # ✅ Always capitalize 'False'. Returns: <class 'bool'>
```

## **Use of Comparison Operators**

```python
1 == 2  # ❌ False → because 1 is not equal to 2
```

```python
2 > 1   # ✅ True → because 2 is greater than 1
```

## **Truthy vs Falsy**

```python
print(bool(0))            # ❌ False → 0 is falsy
print(bool("Python"))     # ✅ True → non-empty string is truthy
print(bool([]))           # ❌ False → empty list is falsy
print(bool([1, 2, 3]))    # ✅ True → non-empty list is truthy
print(bool(None))         # ❌ False → None is always falsy
```

## **Use of "None" keyword**

```python
k = None
print(k)  # Output: None

m = None
print(m)  # Output: None
```
