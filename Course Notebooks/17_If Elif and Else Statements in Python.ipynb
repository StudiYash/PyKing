## **If Statement**

```python
hungry = True
if hungry:
    print("Give me Food")  # ✅ Executed because hungry is True
```

## **If Else Statement**

```python
hungry = False

if hungry:
    print("Give me Food")
else:
    print("Don't give me Food")  # ✅ Executed because hungry is False
```

## **If Elif Statement**

```python
hungry = False
not_hungry = True

if hungry:
    print("Give me Food")
elif not_hungry:
    print("Don't give me Food")  # ✅ This line runs
```

## **If Elif and Else Statement**

```python
starts = False
ends = False

if starts:
    print("We are in danger")
elif ends:
    print("We are out of danger")
else:
    print("We were never in danger")  # ✅ This line runs
```

```python
name = "Yash"

if name == "Jingle":
    print("She is pretty!")
elif name == "James":
    print("He is handsome!")
elif name == "Henry":
    print("He is clever!")
elif name == "Eliot":
    print("He is humble!")
else:
    print("There is no one!")  # ✅ This line runs because none of the conditions matched
```
