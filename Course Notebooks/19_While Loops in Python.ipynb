## **Simple While Loop**

```python
num = 0

while(num < 5):
    print("The number is",num)
    num = num + 1
```

## **While Loop with Else Statement**

```python
num = 0

while(num < 5):
    print("The number is",num)
    num = num + 1
else:
    print("Number is less than 5")
```

## **The pass Keyword**

```python
x = [1,2,3]
for item in x:
    pass
print("pass statement got tested successfully as for loop execution did not took place")
```

## **The continue Keyword**

```python
my_string = "HELLO"
for letter in my_string:
    if letter == "E":
        continue
    print(letter)
```

## **The break Keyword**

```python
x = 0
while x < 5:
    if x == 2:
        break
    print(x)
    x = x + 1
```
