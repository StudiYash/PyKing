## **The .format() Method**

### **First way of using .format() Method :-**

```python
print('This is a string {}'.format('INSERTED'))  # 🔁 Inserts the word inside placeholder → 'This is a string INSERTED'
```

### **Second way of using .format() Method :-**

```python
print('The {} {} {}'.format('fox', 'is', 'running'))  # ➕ Inserts values in the order given
# Output: The fox is running
```

### **Third way of using .format() Method :-**

```python
print('The {f} {i} {r}'.format(f='fox', i='is', r='running'))  # 🎯 Assign each word using named keys
# Output: The fox is running
```

## **The f-strings Method**

### **f-strings method used for single variable**

```python
name = "Yash"
print(f'His name is {name}')  # 🧵 Inserts the variable directly inside the string
```

### **f-strings method used for multiple variables**

```python
name = "Yash"
age = "19"
sex = "boy"

print(f'His name is {name}. His age is {age}. He is a {sex}.')
```

## **Formatting Cheatsheet**

### **Floating Point Precision - `:.2f`**

```python
value = 3.14159

# .format()
print("Pi is {:.2f}".format(value))  # Output: Pi is 3.14

# f-string
print(f"Pi is {value:.2f}")
```

### **Comma Separator - `:,`**

```python
num = 1234567

# .format()
print("Formatted: {:,}".format(num))  # Adds commas → 1,234,567

# f-string
print(f"Formatted: {num:,}")
```

### **Binary Format - `:b`**

```python
num = 42

# .format()
print("Binary: {:b}".format(num))

# f-string
print(f"Binary: {num:b}")
```

### **Hexadecimal Format - `:x`**

```python
num = 255

# .format()
print("Hex: {:x}".format(num))

# f-string
print(f"Hex: {num:x}")
```

### **Octal Format - `:o`**

```python
num = 64

# .format()
print("Octal: {:o}".format(num))

# f-string
print(f"Octal: {num:o}")
```

### **Left Align - `:<10`**

```python
name = "Yash"

# .format()
print("|{:<10}|".format(name))

# f-string
print(f"|{name:<10}|")
```

### **Center Align - `:^10`**

```python
name = "Yash"

# .format()
print("|{:^10}|".format(name))

# f-string
print(f"|{name:^10}|")
```

### **Right Align - `:>10`**

```python
name = "Yash"

# .format()
print("|{:>10}|".format(name))

# f-string
print(f"|{name:>10}|")
```
