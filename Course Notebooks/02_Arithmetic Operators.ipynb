## **Basic Mathematical Operators**

````python
90 + 10  # ➕ Addition Operator: Adds two numbers (90 + 10 = 100)
````

````python
110 - 10 # ➖ Subtraction Operator: Subtracts the second number from the first (110 - 10 = 100)
````

````python
10 * 10  # ✖️ Multiplication Operator: Multiplies two numbers (10 * 10 = 100)
````

````python
1000 / 10  # ➗ Division Operator: Divides first number by second (1000 ÷ 10 = 100.0)
# Note: The result is always a float (decimal), even if the division is exact
````

````python
10 ** 2  # 🔼 Power Operator: 10 raised to the power of 2 = 100
````

````python
10 % 3  # 🧮 Modulus Operator: Returns remainder of 10 divided by 3 → 1
````

## **Combining Mathematical Operators**

````python
10 + 2 - 9 / 3  # ➕➖➗ Mixed operators: Python follows operator precedence (division first)
# 9 / 3 = 3 → 10 + 2 - 3 = 9.0
````

````python
(10 + 2) - (9 / 3)  # 🧠 Parentheses change the order of operations: Still equals 9.0
````

## **FUN FACT**

🎉 **Fun Fact for Curious Minds!**

You might expect:
`0.1 + 0.2 == 0.3`

But Python says:
`0.1 + 0.2 == 0.30000000000000004`

Why? Because of how floating-point numbers are stored in binary.  
Some numbers (like 0.1, 0.2) cannot be exactly represented in binary form, just like 1/3 in decimal.

🛠️ Tip: Use `round()` or `math.isclose()` when comparing floats in real-world programs.

```python
import math
math.isclose(0.1 + 0.2, 0.3)  # Returns True
```

➡️ This is not a bug, but a feature of how computers work!

````python
print(0.1 + 0.2 - 0.3)  # 🤯 Expected output is 0.0, but...
print(0.1 + 0.2)        # ...this shows the internal representation is not exact!
````
