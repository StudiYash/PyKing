# **📘 Notebook 16: Chaining Comparison Operators in Python**

## 🎯 Learning Objectives:
- Chain multiple comparisons together using Python’s shorthand
- Understand how logical operators `and`, `or`, and `not` interact with comparisons
- Improve code readability and efficiency using chained conditions

🕒 Estimated Time: 10–12 minutes

## **Comparison without Operators**

```python
print(2 > 3)        # ❌ False: 2 is not greater than 3
print(2 < 4 > 3)    # ✅ True: 2 < 4 and 4 > 3
```

## **Comparison using AND Logical Operator**

```python
print(2 > 2 and 3 < 3)                   # ❌ False: both are False
print(4 > 3 and "Happy" == "Happy")      # ✅ True: both conditions are True
print(9 >= 8 and "Joker" <= "Joker")     # ✅ True: both sides are True
```

## **Comparison using OR Logical Operator**

```python
print(2 > 2 or 3 < 3)                   # ❌ False: both are False
print(4 > 3 or "Happy" == "Happy")      # ✅ True: at least one is True
print(9 >= 8 or "Joker" <= "Joker")     # ✅ True
```

## **Comparison using NOT Logical Operator**

```python
print(not(2 == 2))  # ❌ False: 2 == 2 is True → not True → False
print(not 3 < 2)    # ✅ True: 3 < 2 is False → not False → True
```

## **Comparison using Multiple Logical Operators**

```python
print(not (3 > 1 and 2 < 5 or 4 == 4))
print(not (5 > 2 and (3 == 3 or 2 != 2) and not (4 <= 4)))
```
