🕒 Estimated Time: 10–12 minutes

## **Variable Assignments to Different Values**

```python
a = 10  # 📦 Assigning the integer 10 to a variable named 'a'
print(a)  # 🖨️ Output the value stored in variable 'a'
# Output: 10
```

```python
b = 20  # 📦 Assigning the integer 20 to variable 'b'
print(b)  # 🖨️ Output the value stored in variable 'b'
# Output: 20
```

## **Combining Multiple Variables**

```python
my_name = "Yash is "
my_income = "Currently Zero"

# ➕ Combining two string variables using + operator
in_short = my_name + my_income

print(in_short)  # Output: Yash is Currently Zero
```

```python
a = a + a  # ♻️ Re-assigning 'a' to be double its current value (10 + 10)
print(a)   # Output: 20
```

## **Use of type() Function**

```python
b = 21
type(b)  # 🔍 Will return: int (since 21 is an integer)
```

```python
c = 54.44
type(c)  # 🔍 Will return: float
```

```python
type("Hello")  # 🔍 Will return: str
```

```python
type(in_short)  # 🔍 Will return: str, since it's a combination of two strings
```
