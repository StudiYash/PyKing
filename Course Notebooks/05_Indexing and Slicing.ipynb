## **Indexing**

### **Normal Indexing**

```python
myname = "Yash Shukla"
print(myname)  # 🖨️ This prints the full string
```

```python
myname[3]  # 🎯 Accessing the 4th character (indexing starts from 0) → 'h'
```

```python
myname[7]  # 🎯 Accessing the 8th character → 'u'
print(myname[7])  # Same as above but using print
```

### **Negative Indexing**

```python
myname[-4]  # 🔄 Accessing from the end: -1 is the last character → 'u'
```

```python
myname[-6]  # → 'S'
print(myname[-6])  # Using print for clarity
```

## **Slicing**

### **Slicing without Step Size Parameter**

```python
myalpha = "abcdefghijkl"
```

```python
myalpha[3:]    # 🧩 Slice from index 3 to the end → 'defghijkl'
```

```python
myalpha[:8]    # 🧩 Slice from start to index 7 → 'abcdefgh'
```

```python
myalpha[3:7]   # 🧩 Slice from index 3 to 6 → 'defg'
```

```python
print(myalpha) # Full string output
```

```python
print(myalpha[::])  # Same as printing the full string
```

### **Step Size parameter**

```python
myalpha
```

```python
myalpha[3:8:2]  # ➿ Picks every 2nd character between index 3 and 7 → 'dfh'
```

### **Python trick to write all charachters in the string in the reverse order**

```python
myalpha[::-1]  # 🔄 Step size of -1 prints string in reverse → 'lkjihgfedcba'
```

### **Negative indexing in slicing**

```python
myalpha[-5:-2]  # 🔁 Extracts part of the string using negative indices → 'hij'
```
