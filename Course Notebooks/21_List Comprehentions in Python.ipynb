## **Normal Operations during List Comprehensions**

### **Creating Lists from Strings**

```python
mystring = "Hello"
mylist = [letter for letter in mystring]
print(mylist)
# → ['H', 'e', 'l', 'l', 'o']
```

```python
mylist2 = [x for x in "MyWord"]
print(mylist2)
# → ['M', 'y', 'W', 'o', 'r', 'd']
```

### **Creating Lists from Ranges**

```python
mylist3 = [x for x in range(0,11)]
print(mylist3)  # [0, 1, 2, ..., 10]
```

## **Complex Operations during List Comprehensions**

### **Applying Functions or Operations**

```python
mylist4 = [num**2 for num in range(0,11)]
print(mylist4)
# → [0, 1, 4, 9, 16, 25, 36, 49, 64, 81, 100]
```

```python
celcius = [0,10,20,30]
farenheit = [((9/5)*temp + 32) for temp in celcius]
print(farenheit)
# → [32.0, 50.0, 68.0, 86.0]
```

### **Filtering Elements**

```python
mylist5 = [num for num in range(0,11) if num%2 == 0]
print(mylist5)
# → [0, 2, 4, 6, 8, 10]
```

### **Applying a Conditional Expression (if...else)**

```python
results = [(x if x%2 == 0 else 'ODD') for x in range(0,11)]
print(results)
# → [0, 'ODD', 2, 'ODD', 4, 'ODD', 6, 'ODD', 8, 'ODD', 10]
```

### **Nested For Loops**

```python
mylist6 = [x*y for x in [2,4,6] for y in [1,10,1000]]
print(mylist6)
# → [2, 20, 2000, 4, 40, 4000, 6, 60, 6000]
```

## **Hidden & Edge Cases: List + Dict Comprehensions**

### **List Comprehension with Nested Ifs (Double filter)**

```python
[x for x in range(20) if x % 2 == 0 if x % 3 == 0]
# → [0, 6, 12, 18]
```

### **List Comprehension with if...else + Filtering (Tricky!)**

```python
[x if x > 5 else 0 for x in range(10) if x % 2 == 0]
# → [0, 0, 0, 6, 8, 0]

# [x if x > 5 for x in range(10) if x % 2 == 0]             # This is invalid
```

### **List of Tuples via Comprehension**

```python
result = [(x, x**2) for x in range(5)]
print(result)
# → [(0, 0), (1, 1), (2, 4), (3, 9), (4, 16)]
```

### **Set Comprehension**

```python
s = {x for x in [1,2,2,3,3,3]}
print(s)
# → {1, 2, 3}
```

### **Nested List Comprehension (2D Flattening)**

```python
matrix = [[1,2,3], [4,5,6], [7,8,9]]
flat = [num for row in matrix for num in row]
print(flat)
# → [1, 2, 3, 4, 5, 6, 7, 8, 9]
```

### **Using enumerate() inside a Comprehension**

```python
indexed = [f"{i}: {v}" for i, v in enumerate(["a", "b", "c"])]
print(indexed)
# → ['0: a', '1: b', '2: c']
```
