## **Creating a Dictionary**

```python
my_dict = {'key1' : 'value1', 'key2' : 'value2'}
print(my_dict)
```

```python
store_prices = {'apple' : '20 rs' , 'orange' : '15 rs' , 'banana' : '18 rs'}
print(store_prices)
```

## **Finding a specific value in the dictionary**

```python
print(my_dict['key1'])

print(store_prices['orange'])
```

## **Dictionaries holding Lists and other Dictioanaries inside them**

```python
dictionary = {'k1':1000 , 'k2':[1,2,3,4,5] , 'k3':{'inside_k1':2000}}
print(dictionary)
```

## **Traversing inside a Dictionary**

```python
print(dictionary['k2'])                 # Access list inside dictionary → [1, 2, 3, 4, 5]
print(dictionary['k2'][1])              # Access 2nd item in list → 2
print(dictionary['k3'])                 # Access nested dictionary → {'inside_k1': 2000}
print(dictionary['k3']['inside_k1'])    # Access nested value → 2000
```

## **Adding new Keys in an already existing Dictionary**

```python
d = {'k1': 1, 'k2': 2}
print(d)

d['k3'] = 3  # ➕ Add a new key
print(d)
```

## **Looking at all Keys inside a Dictionary**

```python
a = {'apple': '20 rs', 'orange': '15 rs', 'banana': '18 rs'}
a.keys()  # 🔑 Returns dict_keys(['apple', 'orange', 'banana'])
```

## **Looking at all Values inside a Dictionary**

```python
a = {'apple': '20 rs', 'orange': '15 rs', 'banana': '18 rs'}
a.values()  # 📦 Returns dict_values(['20 rs', '15 rs', '18 rs'])
```

## **Looking at all Items inside a Dictionary**

```python
a = {'apple': '20 rs', 'orange': '15 rs', 'banana': '18 rs'}
a.items()  # 📚 Returns dict_items([('apple', '20 rs'), ('orange', '15 rs'), ('banana', '18 rs')])
```

## **Using the dict Constructor**

```python
dict_1 = dict([(1,"first"),(2,"Second"),(3,"Third")])
print(dict_1)
```

## **The examples of Complex traversing on dictionaries**

```python
d = {
    'k1': ['a', 'b', 'c'],
    'k2': {'m_1': ['ab', 'cd', 'ef']},
    'k3': {'m_2': {'o_1': ['abc', 'def', 'ghi']}}
}

print((d['k1'][1]).upper())               # B
print((d['k2']['m_1'][1]).upper())        # CD
print((d['k3']['m_2']['o_1'][1]).upper()) # DEF
```
