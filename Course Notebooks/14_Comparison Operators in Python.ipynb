## **Equal To Operator**

```python
print(2 == 2)                  # ✅ True: both integers are equal
print(2 == -1)                 # ❌ False: different values
print('Hello' == 'Hello')      # ✅ True: case-sensitive match
print("Hello" == "hello")      # ❌ False: lowercase ≠ uppercase
print(2 == "Hello")            # ❌ False: int and string are not the same type
print(2 == '2')                # ❌ False: integer 2 is not equal to string '2'
print(3 == 3)                  # ✅ True
```

## **Not Equal To Operator**

```python
print('hELLO' != 'Hello')  # ✅ True: case-sensitive mismatch
print(2 != 1)              # ✅ True
print(2 != 2)              # ❌ False: they are equal
print(2 != '2')            # ✅ True: different data types
```

## **Greater Than Operator**

```python
print(2 > 2)                   # ❌ False
print(2 > 1)                   # ✅ True
print("HELLO" > "hello")       # ❌ False → uppercase < lowercase in Unicode
print('HELLO' > 'HELLO')       # ❌ False → equal strings
```

## **Lesser Than Operator**

```python
print(2 < 2)              # ❌ False
print(1 < 2)              # ✅ True
print("HELLO" < "hello")  # ✅ True → uppercase < lowercase
print('HELLO' < 'HELLO')  # ❌ False
```

## **Greater Than or Equal To Operator**

```python
print(2 >= 2)                # ✅ True
print(2 >= 3)                # ❌ False
print("Yash" >= "yash")      # ❌ False → lowercase is greater
```

## **Lesser Than or Equal To Operator**

```python
print(2 <= 2)                # ✅ True
print(2 <= 3)                # ✅ True
print("Yash" <= "yash")      # ✅ True
```

## **🔎 Summary Table of Comparison Operators**

Assume: `a = 3`, `b = 4`

| Operator | Description                                                        | Example        | Result     |
|----------|--------------------------------------------------------------------|----------------|------------|
| `==`     | True if both operands are equal                                    | `a == b`       | False      |
| `!=`     | True if operands are not equal                                     | `a != b`       | True       |
| `>`      | True if left operand is greater than right                         | `a > b`        | False      |
| `<`      | True if left operand is less than right                            | `a < b`        | True       |
| `>=`     | True if left operand is greater than or equal to right             | `a >= b`       | False      |
| `<=`     | True if left operand is less than or equal to right                | `a <= b`       | True       |
