## **Creating a Set**

````python
my_set1 = set()
print(my_set1)  # This is an empty set → set()
````

````python
my_set2 = set('a')
print(my_set2)  # Creates a set with 1 element: {'a'}
````

## **Common Operations on Set**

### **The `.add()` Method on Set**

````python
my_set2.add(1)
print(my_set2)  # Adds 1 → {'a', 1}
````

### **The `.remove()` Method on Set**

````python
my_set2.remove(1)
print(my_set2)  # Removes 1 → {'a'}
````

````python
my_set2.remove(2) # If we try to remove the value 2 even if it is not present in the set
print(my_set2)  # Python will not ignore it and raise an error as KeyError: 2
````

### **The `.discard()` Method on Set**

````python
my_set2.discard(2)  # Won’t raise error even if 2 isn’t there
````

### **The `.union()` Method on Set**

````python
set_a = {1, 2, 3}
set_b = {3, 4, 5}
print(set_a.union(set_b)) # Union → combines all unique elements
````

### **The `.intersection()` Method on Set**

````python
set_a = {1, 2, 3}
set_b = {3, 4, 5}
print(set_a.intersection(set_b)) # Intersection → common elements
````

### **The `.difference()` Method on Set**

````python
set_a = {1, 2, 3}
set_b = {3, 4, 5}
print(set_a.difference(set_b)) # Difference → elements in a but not the b
print(set_b.difference(set_a)) # Difference → elements in b but not the a
````

### **The `.symmetric_difference()` Method on Set**

````python
set_a = {1, 2, 3}
set_b = {3, 4, 5}
print(set_a.symmetric_difference(set_b)) # Symmetric Difference → elements not shared
````

## **The set() Constructor**

### **The set() Constructor on a List**

````python
my_list = ['a', 'a', 'a', 1, 1, 2, 3, 4, 3, 2, 3, 2]
my_list = set(my_list)
print(my_list)  # Removes duplicates
# Output (order may vary): {'a', 1, 2, 3, 4}
````

### **The set() Constructor on a Tuple**

````python
my_tuple = (1, 2, 3, 1, 8, 2, 'a', 8, 3)
my_tuple = set(my_tuple)
print(my_tuple)
````

### **The set() Constructor on a Dictionary**

````python
my_dict = {'a': 1, 'b': 2, 'c': 3}

# Default: creates set of keys
my_set = set(my_dict)
print(my_set)  # → {'a', 'b', 'c'}

# Get only the values
my_set = set(my_dict.values())
print(my_set)  # → {1, 2, 3}

# Get key-value pairs as tuples
my_set = set(my_dict.items())
print(my_set)  # → {('a', 1), ('b', 2), ('c', 3)}
````

## **Limitations of Sets**

### **The Set cannot be the part of the Set**

````python
set2 = {"Hello", {"World"}}  # ❌ Error: sets cannot be nested inside sets
````

### **Lists can not be a part of the Set**

````python
my_set_2 = {"Hello", [1, 2, 3, 4]}  # ❌ Error: lists are unhashable
print(my_set_2)
````

### **We cannot start a Set by just Adding an Integer**

````python
set([2])  # ✅ Correct way → {2}
set(2)    # ❌ Incorrect → TypeError: int is not iterable
````

## **The `frozenset` — The Immutable Set**

````python
🔒 A `frozenset` is an immutable version of a set:
- Cannot be modified after creation
- Useful as dictionary keys or elements in other sets

✅ You can create one using:
```python
fs = frozenset([1, 2, 3])
```
````

````python
fs = frozenset([1, 2, 3])
fs.add(4)  # ❌ Error: frozenset is immutable → AttributeError
````
