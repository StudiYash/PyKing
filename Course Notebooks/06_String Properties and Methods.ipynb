## **String Concatenation**

```python
name = "Yash"
print(name)  # 🖨️ Print original name
```

```python
new_name = name[1:]  # 🔍 Slice from index 1 onward → 'ash'
print(new_name)
```

```python
name_2 = "H" + new_name  # ➕ Concatenate 'H' with 'ash' → 'Hash'
print(name_2)
```

```python
name = "Yash is "
new_name = name[5:] + "becoming intelligent"  # 'is ' + 'becoming intelligent'
print(new_name)
```

## **String Multipliability**

```python
print(name_2)  # Just to verify → 'Hash'
```

```python
name_3 = name_2 * 2  # 🔁 Repeats the string twice → 'HashHash'
name_3
```

```python
name_2 = "H" + new_name + new_name
print(name_2)
# Output: 'His becoming intelligentis becoming intelligent'
```

```python
name_3 = name_2 * 3 * 4  # Repeats the string 12 times (3 × 4)
print(name_3)
```

## **String Formatting**

### **The .upper() Method**

```python
name_4 = "hello"
name_4.upper()  # 🔠 Converts to uppercase → 'HELLO'
```

```python
name_5 = name_4.upper()
print(name_5)  # Output: HELLO
```

### **The .lower() Method**

```python
name_5 = "HELLO"
name_5.lower()  # 🔡 Converts to lowercase → 'hello'
```

```python
name_6 = name_5.lower()
print(name_6)  # Output: hello
```

### **The .split() Method**

#### **The split method applied directly**

```python
name_7 = "Yash is a stupid boy"
print(name_7)

list_1 = name_7.split()  # 🪓 Splits string into words based on spaces
print(list_1)
# Output: ['Yash', 'is', 'a', 'stupid', 'boy']
```

#### **The split method used to remove a specific letter from the list**

```python
list_1 = name_7.split('s')  # ⚠️ Case-sensitive split on lowercase letter 's'
print(list_1)
# Output: ['Ya', 'h i', ' a ', 'tupid boy']
```
