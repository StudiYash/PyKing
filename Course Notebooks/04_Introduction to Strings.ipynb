## 🎯 Learning Objectives:
- Understand what strings are and how to define them in Python
- Learn how to print strings and use escape characters like `\n` and `\t`
- Use the `len()` function to count characters in a string

🕒 Estimated Time: 12 minutes

## **String Examples**

```python
'hello'           # 🟣 A string wrapped in single quotes
```

```python
"World"           # 🟣 A string wrapped in double quotes
```

```python
"I dont't do that"  # ✅ Works fine even with apostrophes inside double quotes
```

```python
"I have 3 oranges"  # 📄 Strings can contain text and numbers (as characters)
```

## **Printing the Strings**

```python
print('Idiot')  # 🖨️ Prints the string to console
print("mad")    # Works the same with double quotes
```

## **The `\n` ( also called as escape sequence or newline character )**

```python
print("Idiot\nMad")  # 🔄 \n moves the text after it to a new line
```

## **The `\t` ( which is called as tab escape sequence or tab character )**

```python
print("Idiot\tMad")  # 🔄 \t inserts a horizontal tab (like pressing the Tab key)
```

## **The len() Function**

```python
len("Idiot and Mad")  # 🧮 Returns the number of characters including spaces → 13
```

```python
len("Idiot \nMad")  # 🧮 Still counts all characters including `\n` → 10
```
